```python
# n Sum Cols
num_cols = df.select_dtypes('number').columns
print(num_cols)
```

```python
#Numerical Col Analysis
#Kde = Kernal Density Estimator
plt.figure(figsize=(30,10))
first = 1
for i in num_cols:
  plt.subplot(5,5,first)
  plt.title(f'Distribution Analysis by {i}')
  sns.histplot(data=df,x = i,color ='red',kde=True)
  first+=1
plt.show()
```

```python

# text sum column
text_cols = df.select_dtypes('object').columns
print(text_cols)
```

```python
#Numerical Col Analysis
#kde = kernal density Estimator
plt.figure(figsize=(40,20))
first = 1
for i in text_cols:
  plt.subplot(5,2,first)
  sns.countplot(data = df,y = i,color ='skyblue')
  plt.yticks(fontsize = 8)
  first+=1
plt.show()
```

```python
#Top 10 Analysis
for i in text_cols:
  df[i].value_counts().head(10)
  ans = df[i].value_counts().head(10)
  display(ans)
  print('-------------------------------------\n')
```
