# Univariate Analysis

*   Single column Analysis
*  Single textual Col Analysis

*   Single Numerical col Analysis







```python
# n sum cols
num_cols = df.select_dtypes('number').columns
print(num_cols)
```

```python
text_cols = df.select_dtypes('object').columns
print(text_cols)
```

```python
#Numerical Col Analysis
#Kde = Kernal Density Estimator
plt.figure(figsize=(30,10))
first = 1
for i in num_cols:
  plt.subplot(5,5,first)
  plt.title(f'Distribution Analysis by{i}')
  sns.histplot(data =df,x = i ,color = 'red',kde = True)
  first +=1
plt.show()
```

```python
#Textual column Analysis
plt.figure(figsize=(30,10))

first = 1
for i in text_cols:
  plt.subplot(5,5,first)
  sns.countplot(data = df,y = i,color = 'green')
  plt.yticks(fontsize = 8)
  first +=1
plt.show()
```

```python
#Top 10 Analysis
for i in text_cols:
  df[i].value_counts().head(10)
  ans= df[i].value_counts().head(10)
  display(ans)
  print('--------------------------------\n')
```

```python
#Top 10 Analysis
plt.figure(figsize=(30,10))
first = 1
for i in text_cols:
  plt.subplot(5,5,first)
  plt.title(f'Top 10 Analysis of {i}')
  ans= df[i].value_counts().head(10)
  x_value=ans.index
  y_value = ans.values
  chart = sns.barplot(x=x_value,y=y_value,palette=sns.color_palette('cool'))
  for j in chart.containers:
    plt.bar_label(j)

  first += 1
plt.show()
```
